````python
from pathlib import Path
import random
import matplotlib.pyplot as plt
import cv2
import numpy as np

ROOT_DIR = Path.cwd().parent
DATA_DIR = ROOT_DIR / "data" / "processed" / "train"


image_paths = {}
for dir_name in DATA_DIR.iterdir():
    image_paths[dir_name.name] = list(dir_name.glob("*.jpg"))
````

````python
train_paths = (image_paths["Positive"] + image_paths["Negative"])

channel_sum = np.zeros(3, dtype=np.float64)
channel_squared_sum = np.zeros(3, dtype=np.float64)
pixel_count = 0

for image_path in train_paths:
    image = cv2.imread(str(image_path))
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image = image.astype(np.float32) / 255.0
    channel_sum += image.sum(axis=(0,1))
    channel_squared_sum += (image ** 2).sum(axis=(0,1))
    pixel_count += image.shape[0] * image.shape[1]

mean = channel_sum / pixel_count
std = np.sqrt(channel_squared_sum/pixel_count - mean ** 2)

print(mean)
print(std)
````
